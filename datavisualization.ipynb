```python
from numpy import arange
from scipy.stats import norm
from matplotlib import pyplot
x_axis = arange(-3,3,0.01)
y_axis = norm.pdf(x_axis,0.01)
pyplot.plot(x_axis,y_axis)
pyplot.plot(y_axis,x_axis)
pyplot.show
```

```python
from numpy.random import randn
from numpy.random import seed
```

```python
seed(1)
data = 5* randn(10000)  +50
pyplot.hist(data,bins=100)
pyplot.show
```

```python
from numpy import mean
print(mean(data))
```

```python
from numpy import median
print(median(data))
```

```python
from numpy import var
print(var(data))
```

```python
from numpy import sin
from matplotlib import pyplot
x=[x*0.1 for x in range(100)] # consistent intervals for x-axis
y=sin(x)
pyplot.plot(x,y)
pyplot.show()
```

```python
pyplot.bar(x,y)
```

```python
# bar graph
from random import randint
from random import seed
from matplotlib import pyplot
seed(1) # random numbers always start from 1
x=["red","green","yellow"]
y= [randint(0,100),randint(0,100),randint(0,100)]
pyplot.bar(x,y)
pyplot.show()
```
